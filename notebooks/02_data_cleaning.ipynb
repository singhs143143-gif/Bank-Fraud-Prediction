# 🧹 Notebook 02: Data Quality & Cleaning Pipeline
## Data Validation, Deduplication, and Chronological Splitting

This notebook demonstrates rigorous data quality audits, missing value imputation, outlier handling, and time-aware dataset splitting.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

ROOT_DIR = Path.cwd().parent
sys.path.insert(0, str(ROOT_DIR))
from src.config import RAW_DATA_FILE
from src.data_processing import audit_and_clean_data, split_chronological
```

### 1. Execute Data Quality Audits

```python
df_raw = pd.read_csv(RAW_DATA_FILE, low_memory=False)
df_clean = audit_and_clean_data(df_raw)
print(f"Cleaned DataFrame Shape: {df_clean.shape}")
```

### 2. Chronological Splitting (70% Train, 15% Validation, 15% Test)
Transactions are strictly ordered chronologically to prevent temporal data leakage.

```python
train, val, test = split_chronological(df_clean)
print(f"Train Partition : {len(train):,} records | Date range: {train['transaction_timestamp'].min()} to {train['transaction_timestamp'].max()}")
print(f"Val Partition   : {len(val):,} records | Date range: {val['transaction_timestamp'].min()} to {val['transaction_timestamp'].max()}")
print(f"Test Partition  : {len(test):,} records | Date range: {test['transaction_timestamp'].min()} to {test['transaction_timestamp'].max()}")
```

### 3. Verify Split Imbalance Consistency

```python
summary = pd.DataFrame({
    "Partition": ["Train", "Validation", "Test"],
    "Records": [len(train), len(val), len(test)],
    "Fraud Rate (%)": [train["fraud_label"].mean() * 100, val["fraud_label"].mean() * 100, test["fraud_label"].mean() * 100]
})
summary
```
